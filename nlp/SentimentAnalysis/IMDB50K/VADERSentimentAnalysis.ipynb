# VADER Sentiment Analysis

A few words about VADER:
VADER stands for *Valence Aware Dictionary and sEntiment Reasoner*.It is a **lexicon‑ and rule‑based algorithm** for sentiment analysis — **not** a trainable model.

Here’s how it works at a high level:

   1. **Lexicon lookup**
    Each word in the input text is matched against VADER’s dictionary, where it has a valence score from –4 (most negative) to +4 (most positive).

   2. **Rule‑based adjustments**

        - **Negation handling** (e.g., “not good” flips polarity)

        - **Intensity modifiers** (e.g., ALL CAPS or repeated punctuation “GREAT!!!” increase strength)

        - **Contrastive conjunctions** (e.g., “but” shifts emphasis to the clause that follows)

   3. **Output scores**


        VADER returns four metrics:

        `{ neg, neu, pos, compound }`

      
        – where compound is the normalized overall sentiment score between –1 (most negative) and +1 (most positive).

      
        By convention, compound ≥ +0.05 is positive, ≤ –0.05 is negative, and anything in between is neutral.

```python
nltk.download('vader_lexicon')
```

```python
sentiment2label = {"positive": 1, "negative": 0}
label2sentiment = {1: "positive", 0: "negative"}
```

```python
sentiment = SentimentIntensityAnalyzer()
sentiment.polarity_scores("I'm so happy do nlp stuff! :)")
```

```python
compound_binarizer = lambda polarity_scores: 1 if polarity_scores["compound"] >= 0 else 0
```

```python
dataset["sentiment_label"] = dataset["sentiment"].map(sentiment2label)
```

```python
prediction = []
for i, row in tqdm(dataset.iterrows(), total=len(dataset)):
    polarity_scores = sentiment.polarity_scores(row.review)
    prediction.append(compound_binarizer(polarity_scores))
```

```python
vader_accuracy_score = accuracy_score(prediction, dataset["sentiment_label"])
```

```python
print(f"VADER Accuracy: {vader_accuracy_score:.2%}")
```

## Setup Form

```python
text_field = widgets.Text(
    value='',
    placeholder='Type something',
    description='Text:',
    disabled=False   
)
submit_button = widgets.Button(description="Analyze Sentiment")
output = widgets.Output()
def form_fn(b):
    text = text_field.value
    polarity_scores = sentiment.polarity_scores(text)
    sentiment_label = compound_binarizer(polarity_scores)
    sentiment_word = label2sentiment[sentiment_label]
    with output:
        clear_output()
        print(f"Your text has a {sentiment_word} sentiment")
submit_button.on_click(form_fn)
form = widgets.VBox([text_field, submit_button, output])
```
