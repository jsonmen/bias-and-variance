# Import

```python
import nltk
import pandas as pd
from nltk.sentiment import SentimentIntensityAnalyzer
from sklearn.metrics import accuracy_score
from tqdm import tqdm
```

# Dataset

```python
dataset = pd.read_csv("./data/imdb_dataset.csv")
dataset = dataset.head(10000) # Shrink dataset size
```

# VADER Sentiment Analysis

```python
nltk.download('vader_lexicon')
```

```python
sentiment = SentimentIntensityAnalyzer()
sentiment.polarity_scores("I'm so happy do nlp stuff! :)")
```

```python
compound_binarizer = lambda polarity_scores: 1 if polarity_scores["compound"] >= 0 else 0
```

```python
dataset["sentiment_label"] = dataset["sentiment"].map({"positive": 1, "negative": 0})
```

```python
prediction = []
for i, row in tqdm(dataset.iterrows(), total=len(dataset)):
    polarity_scores = sentiment.polarity_scores(row.review)
    prediction.append(compound_binarizer(polarity_scores))
```

```python
vader_accuracy_score = accuracy_score(prediction, dataset["sentiment_label"])
```

```python
print(f"VADER Accuracy: {vader_accuracy_score:.2%}")
```
