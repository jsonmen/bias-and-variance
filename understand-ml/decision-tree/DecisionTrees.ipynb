Adult gnomes are **decision nodes**: they compare feature values to thresholds and direct you left (if the value is less than the threshold) or right (if it's greater or equal). Gnome children are **leaf nodes** that provide the answer. Here's how I do it:  
  1. I take your data and look for the best way to split it into two groups. I go through all features (age, gender, grades) and possible threshold values. **The threshold is chosen as the average between two consecutive values of the feature** to avoid ambiguity.  
  2. For each feature and threshold, I split the data: for example, if the threshold for age is 16.5, students with age < 16.5 go left, and those with age ≥ 16.5 go right.  
  3. In each group, I calculate how well the split helps predict the outcome. For classification tasks (whether lessons are needed), I use **information gain**, which shows how much the split reduces uncertainty. If it were a regression task (e.g., predicting grades), I would minimize the **mean squared error** (MSE).  
  4. I choose the split with the best result and create the first gnome node. Then, I repeat the process for each branch: if the group has enough data (more than the minimum number of samples, say 2), I add another gnome and split further.  
  5. I stop when a branch has too few samples or the tree reaches its maximum depth—these are parameters I set in advance.  
  6. In the leaf nodes (gnome children), for classification, I take the most frequent value (1 or 0), and for regression, the average.

### Train Algorithm

CART (Classification And Regression Trees) - is a  variation of the decision tree algorithm. It can handle both classification and regression tasks. CART was first produced by Leo Breiman, Jerome Friedman, Richard Olshen, and Charles Stone in 1984.


*Quick Note*: In this notebook, I focused on the CART algorithm. There are several training algorithms, such as [ID3](https://en.wikipedia.org/wiki/ID3_algorithm) and [C4.5](https://en.wikipedia.org/wiki/C4.5_algorithm), but CART is used in the scikit-learn library, making it a more practical choice to learn.

- **Step 1**: Start at the Root Node
  - Begin with the entire dataset at the root node.

- **Step 2**: Find the Best Split for Each Feature
  - **Step 2.1**: For every feature in the dataset:
    - **Step 2.1.1**: Sort the unique values of the feature.
    - **Step 2.1.2**: Evaluate potential split thresholds at the midpoints between adjacent sorted values.
    - **Step 2.1.3**:
      - Left child node: Samples with feature values ≤ threshold.
      - Right child node: Samples with feature values > threshold.
  - **Step 2.2**: Calculate the information gain or MSE/variance reduction for each split:
    - Classification: Use Gini impurity or entropy to measure node information gain.
      - Gini: $1 - \sum_{i=1}^{N} p_i^2$, where $p_i$ is the probability of class $i$
      - Entropy: $-\sum_{i=1}^{N} p_i \cdot \log_2(p_i)$, where $p_i$ is the probability of class $i$
      - *Note*: Information Gain = Parent Entropy − (Weighted Average of Child Entropies); can also be done with Gini
    - Regression: Use MSE/variance reduction.
      - Variance Reduction = Parent Variance − (Weighted Average of Child Variances)
      - *Note*: Mean Squared Error (MSE) can also be used instead of variance.

- **Step 3**: Select the Optimal Split
  - Choose the split (feature and threshold) that:
    - Maximizes information gain (for classification).
    - Maximizes variance/MSE reduction (for regression).

- **Step 4**: Split the Node
  - Partition the dataset into left and right child nodes using the selected feature and threshold.

- **Step 5**: Repeat Recursively Until Stopping Criteria Are Met
  - Terminate recursion if any of the following conditions are satisfied:
    - Maximum tree depth reached.
    - Minimum samples per node (e.g., < 5).
    - No significant impurity reduction.

- **Step 6**: Assign Leaf Node Predictions
  - Classification: Majority class in the node.
  - Regression: Mean of target values in the node.

# Decision Tree Classifier

Classification Task: Predict whether a person has cancer based on tumor characteristics.

```python
class Node:
    def __init__(self, 
                 feature_index: int=None, 
                 threshold: float=None, 
                 information_gain: float=None, 
                 value: float=None, 
                 left_node=None, 
                 right_node=None):
        
        # for decision node
        self.feature_index = feature_index # index of feature in dataset
        self.threshold = threshold # condition threshold
        self.left_node = left_node # left decision/leaf node
        self.right_node = right_node # right decision/leaf node
        self.information_gain = information_gain # information gain of split
        
        # for leaf node
        self.value = value # value of leaf node
        
class DecisionTreeClassifier:
    def __init__(self, min_samples_split: int = 2, max_depth: int = 2, criterion="gini"):
        self.root = None # root of the tree
        self.min_samples_split = min_samples_split # min samples that need to split
        self.max_depth = max_depth # max depth of tree
        self.criterion = criterion# criterion string
        self._criterion_func = self._choose_criterion(self.criterion) # criterion function
    
    def _choose_criterion(self, criterion: str):
        """
        Function that take criterion name (string) and return criterion function 
        """
        if criterion == "gini":
            return self._calc_gini_impurity
        elif criterion == "entropy":
            return self._calc_entropy
        else:
            raise ValueError("Incorrect criterion name. Try 'gini' or 'entropy'")
            
    def _build_tree(self, dataset, curr_depth: int=0): 
        """
        Function that recursively build tree
        """
        X, y = dataset[:, :-1], dataset[:, -1]
        num_samples, num_features = X.shape
        if num_samples>=self.min_samples_split and curr_depth < self.max_depth: # Step 5: Repeat Recursively Until Stopping Criteria Are Met
            best_split = self._get_best_split(dataset, num_samples, num_features) # Step 2: Find Best split
            if best_split.get("information_gain", -1) >= 0:
                
                # Step 4: Split the Node
                left_subtree = self._build_tree(best_split["dataset_left"], curr_depth+1)
                right_subtree = self._build_tree(best_split["dataset_right"], curr_depth+1)
                
                return Node(feature_index=best_split["feature_index"],
                            threshold=best_split["threshold"],
                            left_node=left_subtree,
                            right_node=right_subtree, 
                            information_gain=best_split["information_gain"]
                           )
            
        # Step 6: Assign Leaf Node Predictions
        return Node(value=np.mean(y))
    
    def _get_best_split(self, dataset, num_samples, num_features):
        """
        Function that find best split for specific data
        """
        max_information_gain = -float("inf")
        best_split = {}
        for feature_idx in range(num_features): # Step 2.1: For every feature in the dataset
            all_thresholds = dataset[:, feature_idx]
            unique_thresholds = np.unique(all_thresholds) # Step 2.1.1: Sort the unique values of the feature.
            
            if len(unique_thresholds) < 2: # Skip if only one unique value
                continue
            
            midpoints = (unique_thresholds[:-1] + unique_thresholds[1:]) / 2  # Step 2.1.2: Evaluate potential split thresholds at the midpoints between adjacent sorted values.
            
            for threshold in midpoints:
                dataset_left, dataset_right = self._split_dataset(dataset, feature_idx, threshold) # Step 2.1.3: Split dataset by threshold
                if len(dataset_left) > 0 and len(dataset_right) > 0:
                    y, y_left, y_right = dataset[:, -1], dataset_left[:, -1], dataset_right[:, -1]
                    information_gain = self._calc_information_gain(y, y_left, y_right) # Step2.2: Calculate the information gain or mse/variance reduction for each split:
                    if information_gain > max_information_gain: # Step 3: Select the Optimal Split
                        best_split = {
                            "feature_index": feature_idx,
                            "threshold": threshold,
                            "dataset_left": dataset_left,
                            "dataset_right": dataset_right,
                            "information_gain": information_gain,
                        }
                        max_information_gain = information_gain
                    
        return best_split
    def _calc_gini_impurity(self, y):
        count = len(y)
        if count == 0:
            return 0  # Prevent division by zero if empty, though splits should avoid this
        probability_of_one = len(y[y == 1]) / count
        probability_of_zero = 1.0 - probability_of_one
        gini = 1 - (probability_of_one**2 + probability_of_zero**2)
        return gini
        
    def _calc_information_gain(self, y, y_left, y_right) -> float:
        left_weight = len(y_left)/len(y)
        right_weight = len(y_right)/len(y)
        information_gain = self._criterion_func(y) - (left_weight*self._criterion_func(y_left) + right_weight*self._criterion_func(y_right))
        return information_gain
    
    def _split_dataset(self, dataset, feature_index: int, threshold: float): 
        """
        Fuction that split dataset with threshold
        """
        mask = dataset[:, feature_index] <= threshold
        dataset_left = dataset[mask]
        dataset_right = dataset[~mask]
        return (dataset_left, dataset_right)
    
    def fit(self, X, y):
        """
        Function that train decision tree
        """
        X, y = X.copy(), y.copy()
        if isinstance(X, (pd.DataFrame, pd.Series)):
            X = X.to_numpy()
        if isinstance(y, pd.Series):
            y = y.to_numpy()[..., None]
        self._root = self._build_tree(dataset=np.hstack((X, y))) # Step 1: Start from root node
        return self
    
    def _make_prediction(self, x, tree):
        """
        Function that make prediction for 1 value
        """
        if tree.value is not None:
            return tree.value
        if x[tree.feature_index] <= tree.threshold:
            return self._make_prediction(x, tree.left_node)
        else:
            return self._make_prediction(x, tree.right_node)
            
    def predict(self, X):
        """
        Function that make binary prediction for list of values
        """
        preditions = [int(self._make_prediction(x, self._root) > 0.5) for x in X]
        return preditions
    def predict_proba(self, X):
        """
        Function that make probability prediction for list of values
        """
        preditions = [self._make_prediction(x, self._root) for x in X]
        return preditions
    
    def visualize_tree(self, input_column_names: list = None, target_column_name: str = None, probability: bool = False, save_to_file: bool = False, filename: str = "./result/clf_tree", format_: str = "png"):
        """
        Render the binary tree to file <filename>.<format> using Graphviz.
        """
        dot = Digraph(comment="Decision Tree", format=format_)
        node_id = 0

        def add_nodes_edges(node, parent_id=None, edge_label=""):
            nonlocal node_id
            this_id = f"node{node_id}"
            node_id += 1

            # Label: either a decision test or a leaf value
            if node.value is not None:
                label = f"{target_column_name if target_column_name is not None else 'Value'} = {f'{node.value:.2f}' if probability else int(node.value > 0.5)}"
            else:
                label = f"{input_column_names[node.feature_index] if input_column_names is not None else f'X[{node.feature_index}]'} ≤ {node.threshold:.2f}"

            # Add this node
            dot.node(this_id, label=label, shape="box")

            # Add edge from parent
            if parent_id is not None:
                dot.edge(parent_id, this_id, label=edge_label)

            # Recurse
            if node.left_node is not None:
                add_nodes_edges(node.left_node, this_id, edge_label="yes")
            if node.right_node is not None:
                add_nodes_edges(node.right_node, this_id, edge_label="no")

        # Kick off recursion
        add_nodes_edges(self._root)

        # Render to file
        if save_to_file:
            output_path = dot.render(filename, cleanup=True)
            print(f"Tree rendered to: {output_path}")

        return dot
```

# Decion Tree Regressor

Regression Task: Predict depth based on salinity and temperature.

```python
class Node:
    def __init__(self, 
                 feature_index: int=None, 
                 threshold: float=None, 
                 error: float=None, 
                 value: float=None, 
                 left_node=None, 
                 right_node=None):
        
        # for decision node
        self.feature_index = feature_index # index of feature in dataset
        self.threshold = threshold # condition threshold
        self.left_node = left_node # left decision/leaf node
        self.right_node = right_node # right decision/leaf node
        self.error = error # error of split
        
        # for leaf node
        self.value = value # value of leaf node
        
class DecisionTreeRegressor:
    def __init__(self, min_samples_split: int = 7, max_depth: int = 10, criterion: str = "squared_error"):
        self._root = None # root of the tree
        self.min_samples_split = min_samples_split # min samples that need to split
        self.max_depth = max_depth # max depth of tree
        self.criterion = criterion # criterion string
        self._criterion_func = self._choose_criterion(self.criterion) # criterion function
    
    def _choose_criterion(self, criterion: str):
        """
        Function that take criterion name (string) and return criterion function 
        """
        if criterion == "squared_error":
            return self._calc_mse
        elif criterion == "variance_reduction":
            return self._calc_var_red
        else:
            raise ValueError("Incorrect criterion name. Try 'squared_error' or 'variance_reduction'")
            
    def _build_tree(self, dataset, curr_depth: int=0): 
        """
        Function that recursively build tree
        """
        X, y = dataset[:, :-1], dataset[:, -1]
        num_samples, num_features = X.shape
        if num_samples>=self.min_samples_split and curr_depth < self.max_depth: # Step 5: Repeat Recursively Until Stopping Criteria Are Met
            best_split = self._get_best_split(dataset, num_samples, num_features) # Step 2: Find Best split
            if best_split.get("error", -1) >= 0:
                # Step 4: Split the Node
                left_subtree = self._build_tree(best_split["dataset_left"], curr_depth+1)
                right_subtree = self._build_tree(best_split["dataset_right"], curr_depth+1)
                
                return Node(feature_index=best_split["feature_index"],
                            threshold=best_split["threshold"],
                            left_node=left_subtree,
                            right_node=right_subtree, 
                            error=best_split["error"]
                           )
            
        # Step 6: Assign Leaf Node Predictions
        return Node(value=np.mean(y))
    
    def _get_best_split(self, dataset, num_samples, num_features):
        """
        Function that find best split for specific data
        """
        max_error = -float("inf")
        best_split = {}
        for feature_idx in range(num_features): # Step 2.1: For every feature in the dataset
            all_thresholds = dataset[:, feature_idx]
            unique_thresholds = np.unique(all_thresholds) # Step 2.1.1: Sort the unique values of the feature.
            
            if len(unique_thresholds) < 2: # Skip if only one unique value
                continue

            
            midpoints = (unique_thresholds[:-1] + unique_thresholds[1:]) / 2  # Step 2.1.2: Evaluate potential split thresholds at the midpoints between adjacent sorted values.
            
            for threshold in midpoints:
                dataset_left, dataset_right = self._split_dataset(dataset, feature_idx, threshold) # Step 2.1.3: Split dataset by threshold
                if len(dataset_left) > 0 and len(dataset_right) > 0:
                    y, y_left, y_right = dataset[:, -1], dataset_left[:, -1], dataset_right[:, -1]
                    error = self._criterion_func(y, y_left, y_right) # Step2.2: Calculate the information gain or mse/variance reduction for each split:
                    if error > max_error: # Step 3: Select the Optimal Split
                        best_split = {
                            "feature_index": feature_idx,
                            "threshold": threshold,
                            "dataset_left": dataset_left,
                            "dataset_right": dataset_right,
                            "error": error,
                        }
                        max_error = error
        return best_split
    def _calc_mse(self, y, y_left, y_right) -> float:
        """
        Function that calulate MSE of split
        """
        parent_mse = np.mean((y - y.mean())**2)
        n, n_left = len(y), len(y_left)
        n_right = len(y_right)
        mse_left  = np.mean((y_left  - y_left.mean())**2)
        mse_right = np.mean((y_right - y_right.mean())**2)
        weighted_mse = (n_left/n)*mse_left + (n_right/n)*mse_right
        return parent_mse - weighted_mse
    
    def _calc_var_red(self, y, y_left, y_right) -> float:
        """
        Function that calulate Variance Reduction of split
        """
        left_weight = len(y_left) / len(y)
        right_weight = len(y_right) / len(y)
        varince_reduction = np.var(y) - (left_weight * np.var(y_left) + right_weight * np.var(y_right))
        return varince_reduction
    
    def _split_dataset(self, dataset, feature_index: int, threshold: float): 
        """
        Fuction that split dataset with threshold
        """
        mask = dataset[:, feature_index] <= threshold
        dataset_left = dataset[mask]
        dataset_right = dataset[~mask]
        return (dataset_left, dataset_right)
    
    def fit(self, X, y):
        """
        Function that train decision tree
        """
        X, y = X.copy(), y.copy()
        if isinstance(X, (pd.DataFrame, pd.Series)):
            X = X.to_numpy()
        if isinstance(y, pd.Series):
            y = y.to_numpy()[..., None]
        self._root = self._build_tree(dataset=np.hstack((X, y))) # Step 1: Start from root node
        return self
    
    def _make_prediction(self, x, tree):
        """
        Function that make prediction for 1 value
        """
        if tree.value is not None:
            return tree.value
        if x[tree.feature_index] <= tree.threshold:
            return self._make_prediction(x, tree.left_node)
        else:
            return self._make_prediction(x, tree.right_node)
            
    def predict(self, X):
        """
        Function that make prediction for list of values
        """
        preditions = [self._make_prediction(x, self._root) for x in X]
        return preditions
    
    def visualize_tree(self, input_column_names: list = None, target_column_name: str = None, save_to_file: bool = False, filename: str = "./result/reg_tree", format_: str = "png"):
        """
        Render the binary tree to file <filename>.<format> using Graphviz.
        """
        dot = Digraph(comment="Decision Tree", format=format_)
        node_id = 0

        def add_nodes_edges(node, parent_id=None, edge_label=""):
            nonlocal node_id
            this_id = f"node{node_id}"
            node_id += 1

            # Label: either a decision test or a leaf value
            if node.value is not None:
                label = f"{target_column_name if target_column_name is not None else 'Value'} = {node.value:.2f}"
            else:
                label = f"{input_column_names[node.feature_index] if input_column_names is not None else f'X[{node.feature_index}]'} ≤ {node.threshold:.2f}"

            # Add this node
            dot.node(this_id, label=label, shape="box")

            # Add edge from parent
            if parent_id is not None:
                dot.edge(parent_id, this_id, label=edge_label)

            # Recurse
            if node.left_node is not None:
                add_nodes_edges(node.left_node, this_id, edge_label="yes")
            if node.right_node is not None:
                add_nodes_edges(node.right_node, this_id, edge_label="no")

        # Kick off recursion
        add_nodes_edges(self._root)

        # Render to file
        if save_to_file:
            output_path = dot.render(filename, cleanup=True)
            print(f"Tree rendered to: {output_path}")
        return dot
```
